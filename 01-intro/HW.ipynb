```python
#Q7.1
df_asia = df[df["origin"] == "Asia"]
df_asia
```

```python
#Q7.2
df_asia_subset = df_asia[['vehicle_weight', 'model_year']]
df_asia_subset
```

```python
#Q7.3
df_asia_subset.head(7)
```

```python
#Q7.4
X = df_asia_subset.head(7).to_numpy()
X
```

```python
#Q7.5
XTX = X.T @ X
XTX
```

```python
#Q7.6
XTX_inv = np.linalg.inv(XTX)
XTX_inv
```

```python
#Q7.7
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
y
```

```python
#Q7.8
w = XTX_inv @ X.T @ y
w
```

```python
#Q7.9
print("sum of the elements of result = ", w.sum())
```
